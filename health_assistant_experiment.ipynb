# KopanoHealthAI Health Assistant Experiment

## Objective

This notebook explores how artificial intelligence can support health communication and outbreak preparedness.

The experiment is an early prototype of a multilingual AI health assistant designed to improve access to reliable health information in African communities.

## Goals

- Test a basic AI health information system.
- Explore responsible AI applications in healthcare.
- Identify challenges for future multilingual development.

```python
health_information = {
    "malaria": "Malaria prevention includes using mosquito nets, reducing mosquito exposure, and seeking medical advice when symptoms appear.",
    "vaccines": "Vaccines help the immune system recognize and fight specific diseases.",
    "outbreak": "During outbreaks, early detection, accurate information, and rapid response are important."
}


def health_assistant(question):
    question = question.lower()

    for topic, answer in health_information.items():
        if topic in question:
            return answer

    return "Please consult trusted health professionals for accurate medical advice."


question = input("Ask a health question: ")

print(health_assistant(question))
```

```python
# African language health information examples

african_health_information = {
    "xitsonga": {
        "malaria": "Malarhiya i vuvabyi bya switsotswana leswi hangasiwaka hi tinsuna. Tirhisa tinete ta tinsuna ku tisirhelela."
    },
    "sepedi": {
        "malaria": "Bolwetši bja malaria bo bakwa ke ditshenekegi tše di fetšago malwetši. Šomiša malokwa a menoši go itšhireletša."
    }
}


def african_health_assistant(language, topic):
    language = language.lower()
    topic = topic.lower()

    if language in african_health_information:
        if topic in african_health_information[language]:
            return african_health_information[language][topic]

    return "Information not available yet."


print(african_health_assistant("xitsonga", "malaria"))
print(african_health_assistant("sepedi", "malaria"))
```
